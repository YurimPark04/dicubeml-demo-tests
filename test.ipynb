```python
import pandas as pd


def demo_load_prdt_pickle_data(
    user_id: str,
    workspace_id: str,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    print("user_id:", user_id)
    print("workspace_id:", workspace_id)

    train_prdt = pd.DataFrame({
        "id": [1, 2, 3],
        "actual": [100, 200, 300],
        "pred": [110, 190, 310],
    })

    test_prdt = pd.DataFrame({
        "id": [4, 5],
        "actual": [400, 500],
        "pred": [390, 520],
    })

    return train_prdt, test_prdt

```

```python

train_df, test_df = demo_load_prdt_pickle_data(
    user_id="user01",
    workspace_id="workspaceA",
)

print("\n[TRAIN 결과]")
print(type(train_df))

print("\n[TEST 결과]")
print(type(test_df))
```
